```python
# these warnings are fine. you can ignore them.
import random, math
import igraph as ig

import sys
sys.path.append('../')
from util import *

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("Imports finished.")
```

```python
dataset = Dataset(root='/tmp/Cora', name='Cora', device=device)
data, in_feats, h_feats, num_classes = dataset.get_data()

G, x, y, train_mask, test_mask = convert_to_networkx(data)

bbox = (1000, 1000)  # Width and height in pixels
dpi = 300  # Resolution in dots per inch
```

```python
cc = list(nx.strongly_connected_components(G))
nodes = {}
for c in cc:
    print(len(c))
    if len(c) == 26:
        nodes = c
subgraph = G.subgraph(nodes).to_undirected().copy()
print(subgraph.number_of_edges())
```

```python
ig_graph = ig.Graph.from_networkx(subgraph)

# styling
purple_shade = "#040404"
node_width = 20
edge_width = 2

ig_graph.vs["color"] = purple_shade
ig_graph.vs["width"] = node_width
ig_graph.es["width"] = edge_width

layout = ig_graph.layout("kk")

# Plot with igraph and save with a transparent background
fig, ax = plt.subplots()
bbox = (800, 800)  # Define the bounding box size
dpi = 300  # Define the DPI for high resolution

# Plot with igraph
ig.plot(ig_graph, layout=layout, target=ax, bbox=bbox, margin=20)

# Make background transparent
fig.patch.set_alpha(0.0)
ax.patch.set_alpha(0.0)

# Save the figure with a transparent background
plt.savefig("b4ptb.png", transparent=True, bbox_inches='tight', pad_inches=0, dpi=dpi)
plt.close(fig)
```

```python
ig_graph = ig.Graph.from_networkx(subgraph)

purple_shade = "#040404"
node_width = 20
edge_width = 2

ig_graph.vs["color"] = purple_shade
ig_graph.vs["width"] = node_width
ig_graph.es["width"] = edge_width

budget = math.floor(0.15 * ig_graph.ecount())
edges = []

while len(edges) < budget:
    i = random.randint(0, ig_graph.vcount() - 1)
    j = random.randint(0, ig_graph.vcount() - 1)

    if i != j and not ig_graph.are_adjacent(i, j) and not ig_graph.are_adjacent(j, i) and (i, j) not in edges and (j, i) not in edges:
        edges.append((i, j))

ptb_edge_color = "#ec048c"
for e in edges:
    ig_graph.add_edge(e[0], e[1])
    id = ig_graph.get_eid(e[0], e[1])
    ig_graph.es[id]["color"] = ptb_edge_color
    ig_graph.es[id]["width"] = edge_width + 1

# Plot with igraph and save with a transparent background
fig, ax = plt.subplots()
bbox = (800, 800)  # Define the bounding box size
dpi = 300  # Define the DPI for high resolution

# Plot with igraph
ig.plot(ig_graph, layout=layout, target=ax, bbox=bbox, margin=20)

# Make background transparent
fig.patch.set_alpha(0.0)
ax.patch.set_alpha(0.0)

# Save the figure with a transparent background
plt.savefig("afterptb.png", transparent=True, bbox_inches='tight', pad_inches=0, dpi=dpi)
plt.close(fig)
```
