```python
z = [2, 3, 4]

def func(G, x, y, j, new_G, train_mask, test_mask):
    for n in G.nodes():
        sub_nodes = nx.ego_graph(G, n, j).nodes()
    
        for i in sub_nodes:
            if y[n].item() == y[i].item() and i != n:
                add_edge(new_G, n, i, undirected=True)


p.evalFunctionOnModel(Datasets.CORA, Model.GCN, func, z)
```
