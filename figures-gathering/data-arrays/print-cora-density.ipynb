```python
import numpy as np
import matplotlib.pyplot as plt

# Specify the paths to your .npy files
gcn_results_path = './cora/gcn_results_median.npy'
gat_results_path = './cora/gat_results_median.npy'
gsage_results_path = './cora/gsage_results_median.npy'
gsaint_results_path = './cora/gsaint_results_median.npy'
jaccard_results_path = './cora/jaccard_results_median.npy'
ptb_rates_path = './cora/ptb_rates.npy'

# Load the .npy files
gcn_results = np.load(gcn_results_path)
gat_results = np.load(gat_results_path)
gsage_results = np.load(gsage_results_path)
gsaint_results = np.load(gsaint_results_path)
jaccard_results = np.load(jaccard_results_path)
ptb_rates = np.load(ptb_rates_path)

# Define slightly darker pastel colors for the lines
colors = ['#7F58FF', '#FFA500', '#5DBB63', '#FF4C4C', '#FF70D8']  # Slightly darker shades

# Create a plot with a specified size
plt.figure(figsize=(13, 9), facecolor='#163E64')  # Increased width and height
ax = plt.gca()
ax.set_facecolor('#c2dae6')  # Background color inside the axes

# Plot each architecture's results with the slightly darker pastel colors
plt.plot(ptb_rates, gcn_results, marker='o', linestyle='-', color=colors[0], label='GCN')
plt.plot(ptb_rates, gat_results, marker='s', linestyle='-', color=colors[1], label='GAT')
plt.plot(ptb_rates, gsage_results, marker='^', linestyle='-', color=colors[2], label='GSAGE')
plt.plot(ptb_rates, gsaint_results, marker='d', linestyle='-', color=colors[3], label='GSAINT')  # Slightly darker red
plt.plot(ptb_rates, jaccard_results, marker='*', linestyle='-', color=colors[4], label='Jaccard')

# Add title and labels with white color, bold font, and larger font size
plt.title('Change in Accuracy with Classification Density Method', color='white', fontweight='bold', fontsize=20, pad=20)
plt.xlabel('PTB Rates', color='white', fontweight='bold', fontsize=18, labelpad=20)
plt.ylabel('Change in Accuracy', color='white', fontweight='bold', fontsize=18, labelpad=20)

# Customize tick parameters to have white color
ax.tick_params(axis='x', colors='white', labelsize=14)
ax.tick_params(axis='y', colors='white', labelsize=14)

# Add legend with a white background
legend = plt.legend(facecolor='white', fontsize=14)
frame = legend.get_frame()
frame.set_facecolor('white')

# Show grid with light color
plt.grid(color='gray', linestyle='--', linewidth=0.5)

# Adjust the subplot parameters to reduce padding around the axes labels and title
plt.subplots_adjust(left=0.1, right=0.95, top=0.9, bottom=0.1)

# Save plot to a file with higher resolution
plt.savefig('cora_median_new.png', dpi=600, facecolor='#163E64', pad_inches=0.1)

# Show plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Specify the paths to your .npy files
gcn_results_path = './cora/gcn_results_median.npy'
gat_results_path = './cora/gat_results_median.npy'
gsage_results_path = './cora/gsage_results_median.npy'
gsaint_results_path = './cora/gsaint_results_median.npy'
jaccard_results_path = './cora/jaccard_results_median.npy'
ptb_rates_path = './cora/ptb_rates.npy'

# Load the .npy files
gcn_results = np.load(gcn_results_path)
gat_results = np.load(gat_results_path)
gsage_results = np.load(gsage_results_path)
gsaint_results = np.load(gsaint_results_path)
jaccard_results = np.load(jaccard_results_path)
ptb_rates = np.load(ptb_rates_path)

# Define slightly darker pastel colors for the lines
colors = ['#7F58FF', '#FFA500', '#5DBB63', '#FF4C4C', '#FF70D8']  # Slightly darker shades

# Create a plot with a specified size and white background
plt.figure(figsize=(13, 9), facecolor='white')  # Set white background for the figure
ax = plt.gca()
ax.set_facecolor('white')  # Set white background inside the axes

# Plot each architecture's results with the slightly darker pastel colors
plt.plot(ptb_rates, gcn_results, marker='o', linestyle='-', color=colors[0], label='GCN')
plt.plot(ptb_rates, gat_results, marker='s', linestyle='-', color=colors[1], label='GAT')
plt.plot(ptb_rates, gsage_results, marker='^', linestyle='-', color=colors[2], label='GSAGE')
plt.plot(ptb_rates, gsaint_results, marker='d', linestyle='-', color=colors[3], label='GSAINT')  # Slightly darker red
plt.plot(ptb_rates, jaccard_results, marker='*', linestyle='-', color=colors[4], label='Jaccard')

# Add title and labels with black color, bold font, and larger font size
plt.title('Change in Accuracy with Classification Density Method', color='black', fontweight='bold', fontsize=20, pad=20)
plt.xlabel('PTB Rates', color='black', fontweight='bold', fontsize=18, labelpad=20)
plt.ylabel('Change in Accuracy', color='black', fontweight='bold', fontsize=18, labelpad=20)

# Customize tick parameters to have black color
ax.tick_params(axis='x', colors='black', labelsize=14)
ax.tick_params(axis='y', colors='black', labelsize=14)

# Add legend with a white background and black text
legend = plt.legend(facecolor='white', fontsize=14)
frame = legend.get_frame()
frame.set_facecolor('white')
frame.set_edgecolor('black')

# Show grid with a light gray color
plt.grid(color='gray', linestyle='--', linewidth=0.5)

# Adjust the subplot parameters to reduce padding around the axes labels and title
plt.subplots_adjust(left=0.1, right=0.95, top=0.9, bottom=0.1)

# Save plot to a file with higher resolution
plt.savefig('cora_median_new_white_bg.png', dpi=600, facecolor='white', pad_inches=0.1)

# Show plot
plt.show()
```
